# Psychometric analysis for v0.7 and the compact prompt core
Upload processor graph exports. No model calls. Files uploaded here enter your Colab session. The generated report is self-contained, uses no JavaScript, and can be opened locally. The results ZIP includes knowledge_graph.json with typed findings and response evidence. Extract the whole ZIP before opening report.html so its KG download link works; that relative download link is not intended for the inline notebook preview.

```python
from pathlib import Path
import urllib.request, importlib.util
url = "https://raw.githubusercontent.com/dwtrott/llm_psychometric_test/main/analyzer/psychometric.py"
urllib.request.urlretrieve(url, "psychometric.py")
spec = importlib.util.spec_from_file_location("psychometric", "psychometric.py")
psychometric = importlib.util.module_from_spec(spec)
spec.loader.exec_module(psychometric)
```

```python
from google.colab import files
uploaded = files.upload()
paths = list(uploaded)
assert paths, "Upload at least one v0.7 or compact-core graph.json / rated_graph.json export."
report = psychometric.build(paths, "analysis-output")
print("Saved", report)
```

```python
from IPython.display import HTML, display
display(HTML(report.read_text(encoding="utf-8")))
```
